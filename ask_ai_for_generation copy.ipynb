```python
import requests
import json
import pandas as pd
import time
import random
from config import XAI_API_KEY

df_raw_all=pd.read_csv('report_conclusions_raw.csv')
df_sample=df_raw_all[:100]
```

```python
prompt_1="你是一名专业的券商分析师，请参考输入的文本，仿照按照真实券商研究报告的风格，中文撰写一个类似的片段。"
prompt_2="你是一名专业的券商分析师，请参考输入的文本，仿照按照真实券商研究报告的风格，中文撰写一个类似的片段。但不要照搬或者整合改编给你的输入。"


def call_llm(input,prompt=prompt_2):
    url = "https://api.x.ai/v1/chat/completions"
    headers = {
        "Content-Type": "application/json",
        "Authorization": "Bearer "+XAI_API_KEY
    }


    data = {
        "messages": [
            {
                "role": "system",
                "content": "You are a test assistant."
            },
            {
                "role": "user",
                "content": prompt+input
            }
        ],
        "model": "grok-4-latest",
        "stream": False,
        "temperature": 0
    }

    try:
        response = requests.post(url, headers=headers, data=json.dumps(data))
        if response.status_code == 200:
            return response.json()['choices'][0]['message']['content']
        else:
            print(f"API调用失败, 状态码: {response.status_code}")
            return None
    except Exception as e:
        print(f"API调用异常: {e}")
        return None
```
